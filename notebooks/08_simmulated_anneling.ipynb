```python
import numpy as np
import matplotlib.pyplot as plt
```

$$ \min_{x \in \mathbb{R}} f(x) = x^2 + 10 \cdot sin(x)$$

```python
# Definición de la función
def f(x):
    return x**2 + 10*np.sin(x)

# Configuración inicial
current_x = np.random.uniform(-10, 10)
T = 20.0
cooling_rate = 0.95
history = []  # Lista para guardar los puntos explorados

# 3. Ciclo del algoritmo
for i in range(100): # Reducido a 100 iteraciones para que el plot sea legible
    new_x = current_x + np.random.uniform(-1, 1)
    delta = f(new_x) - f(current_x)
    
    # Criterio de aceptación
    if delta < 0 or np.random.rand() < np.exp(-delta / T):
        
        current_x = new_x
        print("iteracion:", i)
        print("current:x",current_x)
    
    # Guardamos el punto actual en cada paso
    history.append(current_x)
    T *= cooling_rate

print("Mínimo aproximado en x =", current_x)

# --- Bloque de Visualización ---

# Crear rango de datos para la curva de la función
x_range = np.linspace(-10, 10, 200)
y_range = f(x_range)

plt.figure(figsize=(10, 6))
plt.plot(x_range, y_range, 'k--', alpha=0.5, label='Función f(x)')
plt.scatter(history, [f(x) for x in history], c=range(len(history)), 
            cmap='viridis', marker='o', s=20, label='Puntos explorados')

# Resaltar el punto final (mínimo hallado)
plt.plot(current_x, f(current_x), 'ro', markersize=10, label='Mínimo final')

plt.colorbar(label='Progreso de la iteración (Tiempo)')
plt.title('Exploración de Recocido Simulado')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```
